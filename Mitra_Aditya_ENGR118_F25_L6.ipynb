## Problem 1

```python
import numpy as np
import matplotlib.pyplot as plt
from numpy.polynomial.polynomial import Polynomial
```

```python
# Given data
lane_width = np.array([2.9, 2.1, 2.3, 2.1, 1.8, 2.7, 1.5, 2.9, 1.5])
distance = np.array([2.4, 1.5, 2.4, 1.8, 1.8, 2.9, 1.2, 3.0, 1.2])
```

```python
plt.figure(figsize=(7,5))
plt.scatter(lane_width, distance, color='blue', label='Data Points')
plt.title("Bike Lane Width vs Distance Between Bikes and Cars")
plt.xlabel("Lane Width (m)")
plt.ylabel("Distance (m)")
plt.grid(True)
plt.legend()
plt.show()
```

```python
# Fit a straight line using least squares
coeffs = np.polyfit(lane_width, distance, 1)
m, b = coeffs  # slope and intercept

# Generate line for plotting
x_fit = np.linspace(min(lane_width), max(lane_width), 100)
y_fit = np.polyval(coeffs, x_fit)
```

```python
# Plot data and regression line
plt.figure(figsize=(7,5))
plt.scatter(lane_width, distance, color='blue', label='Data Points')
plt.plot(x_fit, y_fit, color='red', label=f'Best Fit: y = {m:.2f}x + {b:.2f}')
plt.title("Linear Regression Fit")
plt.xlabel("Lane Width (m)")
plt.ylabel("Distance (m)")
plt.legend()
plt.grid(True)
plt.show()
```

```python
print(f"Slope (m): {m:.3f}")
print(f"Intercept (b): {b:.3f}")
```

```python
safe_distance = 2.0
width_needed = (safe_distance - b) / m
print(f"Required lane width for 2m distance: {width_needed:.2f} m")
```

## Problem 2

```python
x = np.array([5, 8, 10, 13, 15, 20])
y = np.array([15, 18, 22, 27, 30, 31])
```

```python
# Fit models
coeff1 = np.polyfit(x, y, 1)  # First order
coeff2 = np.polyfit(x, y, 2)  # Second order
coeff3 = np.polyfit(x, y, 3)  # Third order
```

```python
# Generate smooth x for plotting
x_model = np.linspace(min(x), max(x), 200)

# Calculate fitted values
y1 = np.polyval(coeff1, x_model)
y2 = np.polyval(coeff2, x_model)
y3 = np.polyval(coeff3, x_model)
```

```python
# Plot all on same figure
plt.figure(figsize=(8,6))
plt.scatter(x, y, color='black', label='Data Points')
plt.plot(x_model, y1, 'r--', label='1st Order (Linear)')
plt.plot(x_model, y2, 'b-', label='2nd Order (Quadratic)')
plt.plot(x_model, y3, 'g-.', label='3rd Order (Cubic)')
plt.title("Polynomial Model Comparison")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True)
plt.show()
```

The third model looks to be the best fit.

## Problem 3

```python
x_data = np.array([0, 1, 2, 3, 4, 5])
y_data = np.array([0, 0.5, 2, 1.5, 1, 1.8])
```

```python
# Fit a third-order regression
coeff3_reg = np.polyfit(x_data, y_data, 3)
y_reg_fit = np.polyval(coeff3_reg, x_data)

plt.figure(figsize=(7,5))
plt.scatter(x_data, y_data, color='blue', label='Data Points')
plt.plot(x_data, y_reg_fit, 'r-', label='3rd Order Regression Fit')
plt.title("Third Order Polynomial Regression")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.legend()
plt.grid(True)
plt.show()
```

```python
print("Regression coefficients:", coeff3_reg)
```
