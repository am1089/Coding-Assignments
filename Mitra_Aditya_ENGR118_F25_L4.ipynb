## Problem 1

```python
import numpy as np
```

```python
# Coefficents
A = np.array([
    [0, -7, 5],
    [0, 4, 7],
    [-4, 3, -7]
])
```

```python
b = np.array([
    [50],
    [-30],
    [40]
])
```

## Problem 3

```python
import numpy as np
from scipy.optimize import root
```

```python
x0 = np.array([0, 1])
```

```python
def f(v):
    x1 = v[0]
    x2 = v[1]
    
    # Equation 1
    f1 = -x1**2 + x1 + 0.75 - x2
    
    # Equation 2
    f2 = x1**2 - x2 - 5 * x1 * x2
    
    # Return an array of the function values
    return np.array([f1, f2])
```

```python
# Find the roots
result = root(f, x0, tol=1e-10)
xsol = result.x
```

```python
print(f"The roots are [x1, x2]: {xsol}")
```
