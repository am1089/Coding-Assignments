## Problem 1

### A

```python
def div_average(a):
    iterations = 1000
    stopping = 0.0000000001
    x = a/2

    if a < 0:
        return "Must be positive"
    elif a == 0:
        return 0
    
    for i in range(iterations):
        next_x = (x + a/x) / 2

        if abs((next_x - x)/next_x)*100 < stopping:
            break
        
        x = next_x
    return x
```

### B

```python
div_average(25)
```

## Problem 3

### A

```python
def New_Raph(z):
    it = 0
    error = 0.000001
    max_iter = 1000
    for i in range(max_iter):
        func = 5*z**4 - 0.5*z**2 - z
        der = 20*z**3 - z
        z_1 = z - (func / der) 
        it += 1
        print(f"Iteration {it}: z = {z_1}")
        if abs((z_1 - z)/z_1)*100 < error:
            return print(f"The final value for z is {z_1}")
        z = z_1
    return print(f"The final value for z is {z}")
```

```python
New_Raph(10)
```
