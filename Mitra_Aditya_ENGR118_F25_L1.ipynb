# Problem 1

**Learn to import libraries (numpy, matplotlib), explore linspace, annotate sample plotting code, and then write your own code to compute and plot a formula**

## 1B

```python
import numpy as np # This imports the numpy library and saves it to be used as np
import matplotlib.pyplot as plt # This imports the pyplot library and saves it to be used as plt
t = np.linspace(0.1,0.25,10) # This creates an array of 10 evenly spaced numbers between 0.1 and 0.25
y = (6*t**3-3*t-4)/8/np.sin(5*t) # This uses the values in t to calculate a y value
plt.plot(t,y,c='r') # This uses the plot function from pyplot and uses t for the x axis and y for the y axis. The color of the line is set to red. 
plt.xlabel('t') # Labels the x axis as t
plt.ylabel('y') # Labels the y axis as y
plt.grid() # Displays the graph and line
```

```python
import numpy as np
import matplotlib.pyplot as plt
t = np.linspace(1,5,5)
y = ((3*t-2)/(4*t)) - (np.pi/2)*t
plt.plot(t,y)
plt.xlabel('t')
plt.ylabel('y')
plt.grid()
```

# Problem 2

**Plot the normal probability density function using linspace from -5 to 5, label axes, and identify the maximum point**

```python
import numpy as np
import math
import matplotlib.pyplot as plt
z = np.linspace(-5,5,100)
y = (1/np.sqrt(2*np.pi)) * math.e**-(z**2/2)
plt.plot(z,y)
plt.xlabel('z')
plt.ylabel('frequency')
plt.grid() 
```
